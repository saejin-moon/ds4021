### Group Discussion Answers

#### a. Key Difference Between PCR and PLS
The key difference is that PCR is unsupervised and PLS is supervised. PCR computes its directions $Z_1, \dots, Z_M$ from the predictors $\mathbf{X}$ alone, picking the linear combinations with the most variance, so the response $Y$ is never used. PLS uses $Y$ too: each component is built to explain predictor variance and at the same time maximize covariance with the response.

#### b. Why PCR Might Fail to Find Directions Useful for Predicting $Y$
PCR assumes the directions where $\mathbf{X}$ varies most are also the ones useful for predicting $Y$, and nothing guarantees that. If the true relationship with $Y$ runs along a low-variance direction (or the biggest components mostly carry noise), PCR drops that direction when it keeps only the first few components. A low-dimensional PCR model can then explain most of the variance in $\mathbf{X}$ and still do no better than predicting the mean.

#### c. How the First Direction $Z_1$ is Computed in PLS
First standardize the $p$ predictors and the response (mean 0, unit variance), as the reading recommends. Then set each loading $\phi_{j1}$ equal to the coefficient from the simple regression of $Y$ onto $X_j$, which is proportional to the correlation between $Y$ and $X_j$ (the covariance, once both are standardized):
   $$\phi_{j1} \propto \text{Cov}(X_j, Y)$$
and $Z_1$ is the normalized combination
   $$Z_1 = \sum_{j=1}^p \phi_{j1} X_j \quad \text{subject to} \quad \sum_{j=1}^p \phi_{j1}^2 = 1$$
So the predictors most correlated with $Y$ on their own get the biggest weights.

#### d. Similarities and Differences Between PLS and Penalized Regression
Both methods solve the same problem: OLS is unstable when $p$ is large and the predictors are correlated. Both accept some bias to cut variance, both use cross-validation to pick their tuning parameter ($M$ components for PLS, $\lambda$/$\alpha$ for Ridge/Lasso/ElasticNet), and both end with a linear model $\hat{Y} = \beta_0 + \sum_{j=1}^p \beta_j X_j$. They differ in how: PLS projects $\mathbf{X}$ onto $M$ components and fits OLS on those; penalized regression instead keeps all $p$ predictors and adds a penalty ($L_2$ for Ridge, $L_1$ for Lasso) to the loss. This matters for sparsity: PLS is dense like Ridge, while Lasso and ElasticNet can shrink coefficients to exactly zero and do feature selection. Penalized regression also solves one convex problem, while PLS builds components one at a time, deflating $\mathbf{X}$ and $Y$ after each.

```python
# step 1: generate synthetic 2d dataset where the target y aligns with the low-variance direction
rng_toy = np.random.RandomState(0)
n_samples_toy = 500
cov_toy = [[3, 3], [3, 4]]
X_toy = rng_toy.multivariate_normal(mean=[0, 0], cov=cov_toy, size=n_samples_toy)

# fit pca on x to find principal directions
pca_raw = PCA(n_components=2).fit(X_toy)

# construct target y strongly correlated with the 2nd principal component (the low-variance direction)
y_toy = X_toy.dot(pca_raw.components_[1]) + rng_toy.normal(size=n_samples_toy) / 2.0

# step 2: split into training (75%) and reporting (25%) partitions
X_train_toy, X_test_toy, y_train_toy, y_test_toy = train_test_split(
    X_toy, y_toy, test_size=0.25, random_state=rng_toy
)

# step 3: fit pcr (pca with 1 component + linear regression) and pls (1 component)
pcr_1comp = make_pipeline(StandardScaler(), PCA(n_components=1), LinearRegression())
pcr_1comp.fit(X_train_toy, y_train_toy)
pca_step = pcr_1comp.named_steps["pca"]

pls_1comp = PLSRegression(n_components=1)
pls_1comp.fit(X_train_toy, y_train_toy)

# step 4: visualize projected data and compare model performance on reporting set
fig_toy, axes_toy = plt.subplots(1, 2, figsize=(11, 4))

# pcr plot
axes_toy[0].scatter(pca_step.transform(X_test_toy), y_test_toy, alpha=0.4, label="Ground Truth", color="tab:blue")
axes_toy[0].scatter(
    pca_step.transform(X_test_toy), pcr_1comp.predict(X_test_toy), alpha=0.4, label="Predictions", color="tab:orange"
)
axes_toy[0].set_xlabel("Projected data onto 1st PCA component")
axes_toy[0].set_ylabel("y")
axes_toy[0].set_title(f"PCR (1 comp) - R²: {pcr_1comp.score(X_test_toy, y_test_toy):.3f}")
axes_toy[0].legend()
axes_toy[0].grid(True, linestyle="--", alpha=0.5)

# pls plot
axes_toy[1].scatter(pls_1comp.transform(X_test_toy), y_test_toy, alpha=0.4, label="Ground Truth", color="tab:blue")
axes_toy[1].scatter(
    pls_1comp.transform(X_test_toy), pls_1comp.predict(X_test_toy), alpha=0.4, label="Predictions", color="tab:orange"
)
axes_toy[1].set_xlabel("Projected data onto 1st PLS component")
axes_toy[1].set_ylabel("y")
axes_toy[1].set_title(f"PLS (1 comp) - R²: {pls_1comp.score(X_test_toy, y_test_toy):.3f}")
axes_toy[1].legend()
axes_toy[1].grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()

print(f"PCR (1 component) Test R²: {pcr_1comp.score(X_test_toy, y_test_toy):.4f}")
print(f"PLS (1 component) Test R²: {pls_1comp.score(X_test_toy, y_test_toy):.4f}")
```

### Step-by-Step Code Explanation

#### Block 1, synthetic data generation
We sample 500 points from a 2D Gaussian with covariance `[[3, 3], [3, 4]]`, so the two features are strongly correlated and PC1 holds most of the variance. Then we build the response on purpose along the second principal component, `y = X @ pca.components_[1] + noise/2`. So the direction with the least variance in `X` is the one that predicts `y`, which is exactly the setup where unsupervised dimension reduction fails.

#### Block 2, training and reporting partition
`train_test_split` sets aside 25% of the samples so every score is computed on data the models never fit. Without this, the numbers would describe the training points and hide overfitting.

#### Block 3, fitting the two 1-component models
Both models must compress two predictors into one direction, and that constraint shows the difference. PCR (PCA + OLS) takes the max-variance direction and misses the signal: its reporting $R^2$ is negative ($\approx -0.03$, worse than predicting the mean). PLS looks at $y$ while building its component, finds the predictive direction, and reaches $R^2 \approx 0.658$ on the same held-out data.

### Reflection on Experiment

The first PLS component is the combination of the two features that maximizes covariance with $y$, and here it almost matches the low-variance principal direction where we placed the signal. Its correlation with $y$ is $|r| \approx 0.811$, so it holds nearly all the predictive information. The second direction follows the reading's recipe: regress each variable on $Z_1$, keep the residuals (the part not explained by the first direction), then build $Z_2$ from those the same way. It barely correlates with $y$ ($|r| \approx 0.191$).

Going from 1 to 2 components helped a little: $R^2$ went from 0.6583 to 0.6732 and MSE from 0.2265 to 0.2166. That is expected with $p = 2$, since the first component already captures the real relationship and the second is mostly noise.

The two methods optimize different objectives. PCA maximizes $\text{Var}(\mathbf{X}v)$ and never uses $y$, so a signal in a low-variance direction gets discarded. PLS maximizes $\text{Cov}^2(\mathbf{X}w, Y) = \text{Var}(\mathbf{X}w) \cdot \text{Corr}^2(\mathbf{X}w, Y) \cdot \text{Var}(Y)$, so each component balances spread against relevance to the response. That is why it recovers the direction PCA discards.

```python
# step 3a: load dataset and prepare partitions
dataset_url = "https://raw.githubusercontent.com/UVADS/DS-4021/84e85842222820f6d60bfdfd051c6d1556d98479/datasets/lab01_data.csv"
try:
    df_mol = pd.read_csv("lab01_data.csv")
except Exception:
    df_mol = pd.read_csv(dataset_url)

# separate molecular fingerprints (x) and permeability target (y)
X_mol = df_mol.drop(columns=["permeability"])
y_mol = df_mol["permeability"]

# split into 70% training and 30% reporting (test) sets with fixed random seed
X_train_mol, X_test_mol, y_train_mol, y_test_mol = train_test_split(
    X_mol, y_mol, test_size=0.30, random_state=42
)

print("=" * 60)
print("Dataset Summary:")
print(f"  • Total Compounds:           {df_mol.shape[0]}")
print(f"  • Total Binary Fingerprints: {X_mol.shape[1]}")
print(f"  • Training Set Size:         {X_train_mol.shape[0]} compounds")
print(f"  • Reporting Set Size:        {X_test_mol.shape[0]} compounds")
print(f"  • Target Permeability Mean:  {y_mol.mean():.2f} (Std: {y_mol.std():.2f}, Range: [{y_mol.min():.2f}, {y_mol.max():.2f}])")
print("=" * 60)
```

```python
# step 3b & 3c: construct pipeline and optimize hyperparameters via cross-validation

# binary variance: p*(1-p). a threshold of 0.01 filters out features present in < 1% or > 99% of samples.
vt_step = VarianceThreshold(threshold=0.01)

# build pls pipeline: variancethreshold -> standardscaler -> plsregression
pls_pipeline = Pipeline([
    ("vt", vt_step),
    ("scaler", StandardScaler()),
    ("pls", PLSRegression()),
])

# 5-fold cross-validation strategy
cv_strategy = KFold(n_splits=5, shuffle=True, random_state=42)

# grid search over number of pls components
param_grid_pls = {
    "pls__n_components": list(range(1, 26))
}

grid_pls = GridSearchCV(
    estimator=pls_pipeline,
    param_grid=param_grid_pls,
    cv=cv_strategy,
    scoring="r2",
    n_jobs=-1,
    return_train_score=True,
)

# train and optimize on training set
grid_pls.fit(X_train_mol, y_train_mol)

# extract required reporting items
vt_fitted = grid_pls.best_estimator_.named_steps["vt"]
n_features_remaining = int(vt_fitted.transform(X_train_mol).shape[1])
n_features_initial = int(X_train_mol.shape[1])
best_n_components = int(grid_pls.best_params_["pls__n_components"])
cv_r2_train = float(grid_pls.best_score_)

print("=" * 65)
print("PLS Pipeline Optimization Results (Training Set):")
print(f"  • Initial number of fingerprint predictors:      {n_features_initial}")
print(f"  • Predictors remaining after VarianceThreshold: {n_features_remaining} (filtered {n_features_initial - n_features_remaining} sparse/near-constant features)")
print(f"  • Optimal number of PLS components:             {best_n_components}")
print(f"  • 5-Fold Cross-Validated Training R²:            {cv_r2_train:.4f}")
print("=" * 65)
```

```python
# step 3c visualization: cross-validation curve for pls components
cv_results = grid_pls.cv_results_
components_range = [p["pls__n_components"] for p in cv_results["params"]]
mean_cv_r2 = cv_results["mean_test_score"]
std_cv_r2 = cv_results["std_test_score"]

fig_cv, ax_cv = plt.subplots(figsize=(9, 4.5))
ax_cv.plot(components_range, mean_cv_r2, marker="o", color="tab:blue", label="Mean CV R²", linewidth=2)
ax_cv.fill_between(
    components_range,
    mean_cv_r2 - std_cv_r2,
    mean_cv_r2 + std_cv_r2,
    alpha=0.15,
    color="tab:blue",
    label="±1 Std Dev CV Band",
)
ax_cv.axvline(best_n_components, color="tab:red", linestyle="--", label=f"Selected M={best_n_components} (R²={cv_r2_train:.3f})")
ax_cv.set_xlabel("Number of PLS Components", fontsize=11)
ax_cv.set_ylabel("Cross-Validated R² Score", fontsize=11)
ax_cv.set_title("PLS Cross-Validation Performance vs. Number of Latent Components", fontsize=12, fontweight="bold")
ax_cv.set_xticks(components_range[::2])
ax_cv.legend(loc="lower right")
ax_cv.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# step 3d: evaluate best pls model on reporting set
best_pls_model = grid_pls.best_estimator_
y_pred_pls_test = best_pls_model.predict(X_test_mol).ravel()

# calculate metrics on reporting set
r2_pls_test = r2_score(y_test_mol, y_pred_pls_test)
mse_pls_test = mean_squared_error(y_test_mol, y_pred_pls_test)
rmse_pls_test = np.sqrt(mse_pls_test)

# plot predicted vs. actual permeability
fig_eval, ax_eval = plt.subplots(figsize=(7, 6))
ax_eval.scatter(
    y_test_mol,
    y_pred_pls_test,
    color="teal",
    edgecolors="k",
    alpha=0.75,
    s=50,
    label="Test Compounds",
)

# reference identity line
min_val = min(y_test_mol.min(), y_pred_pls_test.min()) - 2
max_val = max(y_test_mol.max(), y_pred_pls_test.max()) + 2
ax_eval.plot([min_val, max_val], [min_val, max_val], "r--", linewidth=1.8, label="Ideal Line (y = x)")

ax_eval.set_xlim(min_val, max_val)
ax_eval.set_ylim(min_val, max_val)
ax_eval.set_xlabel("Actual Permeability", fontsize=11)
ax_eval.set_ylabel("Predicted Permeability", fontsize=11)
ax_eval.set_title(
    f"PLS Model Evaluation on Reporting Set\nTest R² = {r2_pls_test:.4f} | Test RMSE = {rmse_pls_test:.2f}",
    fontsize=12,
    fontweight="bold",
)
ax_eval.legend(loc="upper left")
ax_eval.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

print("=" * 60)
print("PLS Reporting Set Performance Evaluation:")
print(f"  • Reporting Set R²:   {r2_pls_test:.4f}")
print(f"  • Reporting Set MSE:  {mse_pls_test:.4f}")
print(f"  • Reporting Set RMSE: {rmse_pls_test:.4f}")
print("=" * 60)
```

### f. Reflect: Discussion on Molecular Permeability Modeling

PLS did the best on the reporting set, with $R^2 = 0.4081$ (MSE $\approx$ 151.57). This fits what the reading says about PLS being widely used in chemometrics (building models from chemical measurements), which is what this fingerprint dataset is. The three penalized PCR models did worse and were close to each other: Ridge 0.3724, Lasso 0.3722, ElasticNet 0.3727, and each one needed 15 principal components plus a penalty. PLS got its result with only 3 components. The reason is how the components are built: PLS picks them while looking at the response, so each one is aimed at predicting permeability; PCA builds its components without ever seeing it.

None of the models is accurate enough to replace a lab permeability assay (Caco-2 or PAMPA). A reporting $R^2$ of about 0.41 means about 59% of the variation in permeability is still unexplained, and an RMSE of about 12.31 is large next to a target that ranges from 0.06 to 55.60. Permeability matters because it decides whether a drug can pass through the intestinal wall or reach the brain, so mistakes this big are costly in both directions: good candidates would be thrown out, and bad ones would move on to expensive testing. The reading also warns that in practice PLS often does no better than ridge regression or PCR, and that its advantage can be a wash. Our gap (0.41 vs about 0.37, on only 50 compounds) is small and comes from one split, which fits that warning.

The case for using the model anyway is different. Computer screening costs almost nothing per molecule and finishes in milliseconds, so it is realistic to rank millions of virtual candidates before doing any chemistry in the lab. A sensible workflow is to let the PLS model narrow down a large library, then test the top compounds for real. The model's limits remain: it is linear over binary fingerprints, so it cannot capture non-linear relationships between structure and permeability, and a small change in a molecule's structure can sometimes cause a big jump in permeability that a linear model will miss.
